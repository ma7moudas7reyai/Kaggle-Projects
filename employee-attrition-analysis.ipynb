 # Employee Attrition Analysis

## Exploring the Factors Behind Employee Turnover

This project analyzes employee data to identify the key factors associated with employee attrition and understand the characteristics of employees who leave the company.

The analysis covers employee demographics, job characteristics, satisfaction levels, income, overtime, business travel, and years of experience to uncover meaningful patterns in employee turnover.

 ## Project Objective

The main objective of this project is to explore employee attrition and identify the factors associated with employees leaving the company.

### Key Questions

- What is the overall employee attrition rate?
- Is overtime associated with higher attrition?
- How does job satisfaction relate to employee turnover?
- Which job roles have the highest attrition?

## 3. Exploratory Data Analysis

In this section, we explore the main factors associated with employee attrition using interactive visualizations and descriptive analysis.

### Employee Attrition Overview

We first examine the overall distribution of employee attrition to understand how common employee turnover is in the dataset.

```python
attrition_counts = df["Attrition"].value_counts().reset_index()
attrition_counts.columns = ["Attrition", "Count"]

fig = px.bar(
    attrition_counts,
    x = "Attrition",
    y = "Count",
    title = "Employee Attrition"
)

fig.show()
```

# ****The majority of employees stayed with the company, while a smaller proportion left, indicating that employee attrition is relatively limited compared to employee retention.****

```python
overtime_attrition = (df.groupby("OverTime")["Attrition"].value_counts(normalize = True).mul(100).rename("Percentage").reset_index())

overtime_attrition = overtime_attrition[overtime_attrition["Attrition"] == "Yes"]

overtime_attrition
```

### Attrition Rate by Overtime

We examine whether working overtime is associated with a higher employee attrition rate.

```python
fig = px.bar(
    overtime_attrition,
    x = "OverTime",
    y = "Percentage",
    text = "Percentage",
    title = "Attrition Rate by Overtime"
)

fig.update_traces(
    texttemplate = "%{text:.1f}%",
    textposition = "outside"
)

fig.update_layout(
    yaxis_title = "Attrition Rate (%)",
    xaxis_title = "OverTime"
)

fig.show()
```

# ****Employees who work overtime have a noticeably higher attrition rate than employees who do not, suggesting that overtime may be an important factor associated with employee turnover.****

```python
satisfaction_attrition = (df.groupby("JobSatisfaction")["Attrition"].value_counts(normalize = True).mul(100).rename("Percentage").reset_index())

satisfaction_attrition = satisfaction_attrition[satisfaction_attrition["Attrition"] == "Yes"]

satisfaction_attrition
```

### Attrition Rate by Job Satisfaction

We analyze whether employee job satisfaction is associated with employee turnover.

```python
fig = px.bar(
    satisfaction_attrition,
    x = "JobSatisfaction",
    y = "Percentage",
    text = "Percentage",
    title = "Attrition Rate by Job Satisfaction"
)

fig.update_traces(
    texttemplate = "%{text:.1f}%",
    textposition = "outside"
)

fig.update_layout(
    xaxis_title = "Job Satisfaction Level",
    yaxis_title = "Attrition Rate (%)"
)

fig.show()
```

# Insight: Employees with the lowest job satisfaction (Level 1) have the highest attrition rate at 22.8%, while employees with the highest satisfaction (Level 4) have the lowest attrition rate at 11.3%. This suggests that lower job satisfaction is associated with a higher likelihood of employee turnover.# 

```python
role_attrition = (df.groupby("JobRole")["Attrition"].value_counts(normalize = True).mul(100).rename("Percentage").reset_index())

role_attrition = role_attrition[role_attrition["Attrition"] == "Yes"].sort_values("Percentage", ascending = False)

role_attrition
```

### Attrition Rate by Job Role

We compare attrition rates across different job roles to identify roles with higher employee turnover.

```python
fig = px.bar(
    role_attrition,
    x = "Percentage",
    y = "JobRole",
    orientation="h",
    text = "Percentage",
    title = "Attrition Rate by Job Role"
)

fig.update_traces(
    texttemplate = "%{text:.1f}%",
    textposition = "outside"
)

fig.update_layout(
    xaxis_title = "Attrition Rate (%)",
    yaxis_title = "Job Role"
)

fig.show()
```

# Insight: Attrition varies significantly across job roles. Sales Representatives have the highest attrition rate at 39.8%, followed by Laboratory Technicians (23.9%) and Human Resources (23.1%). In contrast, Research Directors have the lowest attrition rate at only 2.5%. This suggests that employee turnover is strongly associated with job role.

### Monthly Income and Attrition

We compare the distribution of monthly income between employees who left the company and those who stayed.

```python
fig = px.box(
    df,
    x = "Attrition",
    y = "MonthlyIncome",
    title = "Monthly Income Distribution by Attrition",
    points = False
)

fig.update_layout(
    xaxis_title = "Attrition",
    yaxis_title = "Monthly Income"
)

fig.show()
```

# Insight: Employees who left the company generally had lower monthly incomes than those who stayed. The median monthly income for employees who left is around $3.2K, compared with approximately $5.2K for employees who stayed, suggesting that lower income is associated with higher employee attrition.

### Age and Attrition

We compare employee age distributions between employees who left and those who remained with the company.

```python
fig = px.box(
    df,
    x = "Attrition",
    y = "Age",
    title = "Age Distribution by Attrition",
    points = False
)

fig.update_layout(
    xaxis_title = "Attrition",
    yaxis_title = "Age"
)

fig.show()
```

# Insight: Employees who left the company tend to be younger, with a median age of approximately 32 years, compared with around 36 years for employees who stayed. This suggests that younger employees are more likely to leave the company.

```python
bins = [0, 2, 5, 10, 20, 50]
labels = ["0-2", "3-5", "6-10", "11-20", "21+"]

df["YearsAtCompanyGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins = bins,
    labels = labels,
    include_lowest = True
)
```

```python
years_attrition = (df.groupby("YearsAtCompanyGroup", observed = False)["Attrition"].value_counts(normalize = True).mul(100).rename("Percentage").reset_index())

years_attrition = years_attrition[years_attrition["Attrition"] == "Yes"]
```

### Attrition Rate by Years at Company

We examine whether employee tenure is associated with different levels of attrition.

```python
fig = px.bar(
    years_attrition,
    x = "YearsAtCompanyGroup",
    y = "Percentage",
    text ="Percentage",
    title = "Attrition Rate by Years at Company"
)

fig.update_traces(
    texttemplate = "%{text:.1f}%",
    textposition = "outside"
)

fig.update_layout(
    xaxis_title = "Years at Company",
    yaxis_title = "Attrition Rate (%)"
)

fig.show()
```

# Insight: Employees who have been with the company for 0–2 years have the highest attrition rate at 29.8%, while employees with 11–20 years have the lowest rate at 6.7%. This suggests that employees are most likely to leave during their early years at the company.

```python
travel_attrition = (df.groupby("BusinessTravel")["Attrition"].value_counts(normalize = True).mul(100).rename("Percentage").reset_index())

travel_attrition = travel_attrition[travel_attrition["Attrition"] == "Yes"]
```

### Attrition Rate by Business Travel

We investigate whether the frequency of business travel is associated with employee attrition.

```python
fig = px.bar(
    travel_attrition,
    x = "BusinessTravel",
    y = "Percentage",
    text = "Percentage",
    title = "Attrition Rate by Business Travel"
)

fig.update_traces(
    texttemplate = "%{text:.1f}%",
    textposition = "outside"
)

fig.update_layout(
    xaxis_title = "Business Travel",
    yaxis_title = "Attrition Rate (%)"
)

fig.show()
```

# Insight: Employees who travel frequently for business have the highest attrition rate at 24.9%, compared with 15.0% for employees who travel rarely and only 8.0% for non-traveling employees. This suggests that frequent business travel is associated with a higher likelihood of employee turnover.

```python
df_corr = df.copy()

df_corr["Attrition"] = df_corr["Attrition"].map({"Yes": 1, "No": 0})

corr = df_corr.select_dtypes(include="number").corr()
```

### Correlation Analysis

We examine correlations between numerical variables to identify relationships that may be associated with employee attrition.

```python
fig = px.imshow(
    corr,
    text_auto = ".2f",
    aspect = "auto",
    title = "Correlation Heatmap"
)

fig.update_layout(
    width = 1000,
    height = 900
)

fig.show()
```

# Insight: Attrition shows weak negative correlations with several employee characteristics, including TotalWorkingYears (-0.17), JobLevel (-0.17), Age (-0.16), and MonthlyIncome (-0.16). This indicates that employees with greater experience, higher job levels, older ages, and higher incomes tend to have slightly lower attrition. However, the correlations are relatively weak, suggesting that no single numerical factor strongly explains employee turnover.

```python
role_overtime_attrition = (df.groupby(["JobRole", "OverTime"])["Attrition"].apply(lambda x: (x == "Yes").mean() * 100).reset_index(name = "AttritionRate"))
```

### Attrition Rate by Job Role and Overtime

We combine job role and overtime to investigate whether the relationship between overtime and attrition varies across different job roles.

```python
fig = px.bar(
    role_overtime_attrition,
    x = "JobRole",
    y = "AttritionRate",
    color = "OverTime",
    barmode = "group",
    text = "AttritionRate",
    title = "Attrition Rate by Job Role and Overtime"
)

fig.update_traces(
    texttemplate = "%{text:.1f}%",
    textposition = "outside"
)

fig.update_layout(
    xaxis_title = "Job Role",
    yaxis_title = "Attrition Rate (%)",
    xaxis_tickangle = -45
)

fig.show()
```

# Insight: Overtime is associated with substantially higher attrition in several job roles. The strongest differences appear among Sales Representatives (66.7% vs. 28.8%), Laboratory Technicians (50.0% vs. 15.7%), and Human Resources (38.5% vs. 17.9%) for employees working overtime versus those who do not. However, this pattern is not consistent across every role, indicating that the relationship between overtime and attrition may vary by job role.

## Conclusion

The analysis reveals several important patterns associated with employee attrition.

- Overtime is associated with higher employee attrition.
- Lower job satisfaction is associated with higher turnover.
- Sales Representatives have the highest overall attrition rate among job roles.
- Employees in their first 2 years at the company show the highest attrition rate.
- Frequent business travelers have higher attrition compared with non-travelers.
- Employees who leave tend to be younger and have lower monthly incomes.
- The combination of job role and overtime highlights particularly high attrition in some roles.

Overall, employee attrition appears to be influenced by a combination of job satisfaction, workload, tenure, compensation, business travel, and job role rather than a single factor.

These findings can help HR teams identify groups with higher turnover risk and investigate potential improvements in employee retention.
