## App Distribution by Category

Which app categories have the largest number of applications in the Google Play Store dataset?

```python
category_counts = df['Category'].value_counts().reset_index()
category_counts.columns = ['Category', 'App Count']

fig = px.bar(
    category_counts,
    x = 'App Count',
    y = 'Category',
    orientation = 'h',
    title = 'Number of Apps by Category',
    labels = {
        'App Count': 'Number of Apps',
        'Category': 'Category'
    }
)

fig.update_layout(yaxis = {'categoryorder': 'total ascending'})

fig.show()
```

### Insight

The FAMILY category has the highest number of apps by a large margin, followed by GAME and TOOLS. Most other categories contain considerably fewer apps, indicating that the app market in this dataset is heavily concentrated in a few major categories.

## Free vs Paid Apps

What is the distribution of free and paid applications in the Google Play Store dataset?

```python
type_counts = df['Type'].value_counts().reset_index()
type_counts.columns = ['Type', 'App Count']

fig = px.pie(
    type_counts,
    names = 'Type',
    values = 'App Count',
    hole = 0.5,
    title = 'Free vs Paid Apps',
)

fig.show()
```

### Insight

Free apps dominate the Google Play Store dataset, accounting for approximately 92.6% of all apps, while paid apps represent only 7.4%. This indicates that the free-to-download model is far more common than the paid model in this dataset.

## App Rating Distribution

How are app ratings distributed across applications, and what rating range is most common?

```python
fig = px.histogram(
    df.dropna(subset = ['Rating']),
    x = 'Rating',
    nbins = 20,
    title = 'Distribution of App Ratings',
    labels = {'Rating': 'Rating'}
)

fig.show()
```

### Insight

App ratings are heavily concentrated at the higher end of the scale, with most applications receiving ratings between 4.0 and 4.7. Ratings below 3.0 are relatively uncommon, indicating that highly rated apps dominate the dataset.

## Reviews vs Installs

Is there a relationship between the number of installs and the number of user reviews?

```python
fig = px.scatter(
    df,
    x = 'Installs',
    y = 'Reviews',
    title = 'Reviews vs Installs',
    labels = {
        'Installs': 'Number of Installs',
        'Reviews': 'Number of Reviews'
    },
    log_x = True,
    log_y = True,
    hover_data = ['App', 'Category']
)

fig.show()
```

### Insight

There is a strong positive relationship between the number of installs and user reviews. Apps with higher install counts generally receive substantially more reviews, although the number of reviews varies considerably among apps with similar install levels.

## Most Installed Apps

Which applications have the highest number of installs in the Google Play Store dataset?

```python
top_apps = (
    df.groupby('App', as_index = False)['Installs']
    .max()
    .sort_values('Installs', ascending = False)
    .head(10)
)

fig = px.bar(
    top_apps.sort_values('Installs'),
    x = 'Installs',
    y = 'App',
    orientation = 'h',
    title = 'Top 10 Most Installed Apps',
    labels = {
        'Installs': 'Number of Installs',
        'App': 'App'
    }
)

fig.show()
```

### Insight

The top 10 most installed apps all appear in the 1 billion+ install category, showing that these apps have reached the highest install tier recorded in the dataset. This also highlights a limitation of the `Installs` variable, as the values are grouped into install ranges rather than exact download counts.

## User Sentiment Distribution

What is the overall sentiment of users toward the apps based on their reviews?

```python
sentiment_counts = sentiment_df['Sentiment'].value_counts().reset_index()
sentiment_counts.columns = ['Sentiment', 'Review Count']

fig = px.pie(
    sentiment_counts,
    names = 'Sentiment',
    values = 'Review Count',
    hole = 0.5,
    title = 'User Sentiment Distribution'
)

fig.show()
```

### Insight

Positive reviews account for 64.0% of the analyzed reviews, making them the dominant sentiment. Negative reviews represent 21.3%, while neutral reviews account for 14.7%, indicating that user feedback is generally more positive than negative in this dataset.

## App Rating vs User Sentiment

Do apps with higher Google Play ratings tend to receive more positive user reviews?

```python
sentiment_summary = (
    sentiment_df
    .groupby('App')['Sentiment']
    .value_counts(normalize = True)
    .unstack(fill_value = 0)
    .reset_index()
)

sentiment_summary.head()
```

```python
app_analysis = df[['App', 'Rating']].dropna(subset=['Rating']).drop_duplicates('App')

app_analysis = app_analysis.merge(
    sentiment_summary,
    on='App',
    how='inner'
)

app_analysis.head()
```

```python
fig = px.scatter(
    app_analysis,
    x = 'Rating',
    y = 'Positive',
    title = 'App Rating vs Positive Review Sentiment',
    labels = {
        'Rating': 'App Rating',
        'Positive': 'Positive Review Percentage'
    },
    hover_data = ['App']
)

fig.update_yaxes(tickformat = '.0%')

fig.show()
```

### Insight

Apps with higher ratings generally tend to have a higher percentage of positive reviews. However, the wide spread of points at most rating levels shows that app ratings do not always align closely with review sentiment, suggesting that other factors may influence users' written feedback.

## User Sentiment by App Category

Do user sentiments differ across different app categories?

```python
# Merge reviews with app categories
reviews_with_category = sentiment_df.merge(
    df[['App', 'Category']],
    on = 'App',
    how = 'inner'
)

# Select top 10 categories by number of reviews
top_categories = (
    reviews_with_category['Category']
    .value_counts()
    .head(10)
    .index
)

category_sentiment = (
    reviews_with_category[
        reviews_with_category['Category'].isin(top_categories)
    ]
    .groupby('Category')['Sentiment']
    .value_counts(normalize = True)
    .unstack(fill_value = 0)
    .reset_index()
)

category_sentiment
```

```python
fig = px.bar(
    category_sentiment,
    x = 'Category',
    y = ['Positive', 'Negative', 'Neutral'],
    title = 'User Sentiment by App Category',
    labels = {
        'value': 'Percentage of Reviews',
        'Category': 'App Category',
        'variable': 'Sentiment'
    },
    barmode = 'group'
)

fig.update_yaxes(tickformat = '.0%')

fig.show()
```

### Insight

Positive sentiment dominates across all of the top 10 app categories. Health & Fitness has the highest share of positive reviews at approximately 77%, while Game has the highest share of negative reviews at about 38%. This suggests that user satisfaction varies noticeably across categories, with some categories receiving more favorable feedback than others.
